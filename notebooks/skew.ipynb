```python
from pyspark.sql.functions import *

skew_df = (
    spark.range(100000)
    .withColumn(
        "region",
        when(col("id") < 90000, "Dallas")
        .otherwise("Austin")
    )
)

display(
    skew_df.groupBy("region").count()
)
```

```python
from pyspark.sql.functions import rand, floor

salted_df = (
    skew_df
    .withColumn(
        "salt",
        floor(rand() * 5)
    )
)
```
