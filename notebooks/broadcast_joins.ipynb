```python
from pyspark.sql import Row

region_lookup = spark.createDataFrame([
    Row(region="Dallas", state="TX"),
    Row(region="Austin", state="TX"),
    Row(region="Houston", state="TX"),
    Row(region="Chicago", state="IL"),
    Row(region="New York", state="NY"),
    Row(region="Atlanta", state="GA")
])

display(region_lookup)
```

```python
silver_df = spark.table(
    "workspace.telecom_silver.network_events"
)

joined_df = silver_df.join(
    region_lookup,
    "region",
    "left"
)

display(joined_df.limit(10))
```

```python
from pyspark.sql.functions import broadcast

broadcast_df = silver_df.join(
    broadcast(region_lookup),
    "region",
    "left"
)

display(broadcast_df.limit(10))
```

```python
broadcast_df.explain(True)
```
