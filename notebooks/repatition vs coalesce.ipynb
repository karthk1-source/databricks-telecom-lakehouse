```python
bronze_df = spark.table("workspace.telecom_bronze.network_events")

repart_df = bronze_df.repartition(8)

repart_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.telecom_bronze.network_events_repartition_demo")
```

```python
coalesce_df = bronze_df.coalesce(2)

coalesce_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.telecom_bronze.network_events_coalesce_demo")
```
