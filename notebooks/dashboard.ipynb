```python
%sql
SELECT
  region,
  ROUND(AVG(avg_latency_ms), 2) AS avg_latency_ms
FROM workspace.telecom_gold.network_kpi_daily
GROUP BY region
ORDER BY avg_latency_ms DESC;
```

```python
%sql
SELECT
  region,
  ROUND(AVG(dropped_call_pct), 2) AS dropped_call_pct
FROM workspace.telecom_gold.network_kpi_daily
GROUP BY region
ORDER BY dropped_call_pct DESC;
```

```python
%sql
SELECT
  network_type,
  SUM(total_events) AS total_events,
  ROUND(AVG(avg_latency_ms), 2) AS avg_latency_ms,
  ROUND(AVG(dropped_call_pct), 2) AS dropped_call_pct
FROM workspace.telecom_gold.network_kpi_daily
GROUP BY network_type
ORDER BY total_events DESC;
```
